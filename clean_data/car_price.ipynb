### Load data using CSV
---

```python
import pandas as pd
# Replace 'dataset.csv' with the path to your dataset
data = pd.read_csv('cleaned_file.csv')

# Display first few rows
print(data.head())
print(f"Total Row: {data.shape[0]}")
```

---
### Drop duplicate data
---

```python
# Drop duplicate rows from the dataset
data = data.drop_duplicates()

# Display the sizes of the original and cleaned datasets
print(f"Original dataset size: {data.shape[0]}")
print(f"Dataset size after removing duplicates: {data.shape[0]}")

# Display the cleaned dataset
print("\nCleaned Dataset (No Duplicates):")
print(data)
```

---
### Drop NA value in column
---

```python
columns_to_check = ['Body Type', 'Model', 'Year', 'Price', 'City', 'Fuel', 'Transmission', 'Color', 'Condition']

# Drop rows where any specified column is null
print("Original Data:", data.shape[0])
data = data.dropna(subset=columns_to_check)

# Display the cleaned dataset
print("\nCleaned Dataset: ", data.shape[0])
```

---
### Check outlier 
---

```python
def remove_outliers_by_group(df, group_columns, target_column):
    """
    Remove outliers from the target_column within each group defined by group_columns.
    """
    # Initialize an empty list to store cleaned groups
    cleaned_groups = []

    # Iterate over each group
    for _, group in df.groupby(group_columns):
        q1 = group[target_column].quantile(0.25)
        q3 = group[target_column].quantile(0.75)
        iqr = q3 - q1
        lower_bound = q1 - 1.5 * iqr
        upper_bound = q3 + 1.5 * iqr
        # Filter the group for values within the bounds
        cleaned_group = group[(group[target_column] >= lower_bound) & (group[target_column] <= upper_bound)]
        cleaned_groups.append(cleaned_group)

    # Concatenate all cleaned groups back into a single DataFrame
    return pd.concat(cleaned_groups, ignore_index=True)

# Apply the updated function
data_cleaned = remove_outliers_by_group(data, ['Brand', 'Model'], 'Price')

print(f"Original dataset size: {data.shape[0]}")
print(f"Cleaned dataset size: {data_cleaned.shape[0]}")
print(data_cleaned.head())
```
