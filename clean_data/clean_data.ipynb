## 1. Load data from CSV

```python
import pandas as pd

# Step 1: Read the CSV file
csv_file_path = '../scraping_data/khmer24_cars.csv'  # Replace with your actual file path
data = pd.read_csv(csv_file_path)

# Step 2: Drop duplicate rows
# If you want to consider all columns for duplicates
cleaned_data = data.drop_duplicates()

# If you want to consider specific columns for duplicates
# cleaned_data = data.drop_duplicates(subset=['Column1', 'Column2'])

# Step 3: Save the cleaned data back to a CSV file (optional)
cleaned_csv_file_path = 'cleaned_file.csv'
cleaned_data.to_csv(cleaned_csv_file_path, index=False)

# Print the first few rows of the cleaned data
print(cleaned_data.head())

# Print the number of rows before and after dropping duplicates
print(f"Rows before cleaning: {len(data)}")
print(f"Rows after cleaning: {len(cleaned_data)}")
```

## 2. Drop row if Body Type Null

```python
import pandas as pd

# Step 1: Load the CSV file
csv_file_path = 'cleaned_file.csv'  # Replace with the actual file path
data = pd.read_csv(csv_file_path)

# Step 2: Drop rows where any specified column is null
columns_to_check = ['Body Type', 'Model', 'Year', 'Price', 'City', 'Fuel', 'Transmission', 'Color', 'Condition']
cleaned_data = data.dropna(subset=columns_to_check)

# Step 3: Save the cleaned data back to a new CSV file (optional)
cleaned_csv_file_path = 'final_cleaned_data.csv'
cleaned_data.to_csv(cleaned_csv_file_path, index=False)

# Step 4: Print the first few rows of the cleaned data
print(cleaned_data.head())

# Print the number of rows before and after cleaning
print(f"Rows before cleaning: {len(data)}")
print(f"Rows after cleaning: {len(cleaned_data)}")
```
