```python
import requests
from bs4 import BeautifulSoup
import csv

# URL of the website
url = "https://www.khmer24.com/en/c-cars-for-sale?keyword=&ad_field=lexus&ad_type=other"

# Send a request to the website
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/117.0.0.0 Safari/537.36'
}
response = requests.get(url, headers=headers)
html_content = response.content

# Parse the HTML content
soup = BeautifulSoup(html_content, 'html.parser')

# Find all car listings
car_listings = soup.find_all('article', class_='width_list_grid_dt')
print(f"Found {len(car_listings)} car listings.")  # Debugging

# Initialize a list to store the data
data = []

# Loop through each car listing and extract the required information
for listing in car_listings:
    try:
        # Extract car ID from the href link
        link_tag = listing.find('a', class_='position_relative d_block')
        car_id = link_tag.get('href', '').split('-')[-1] if link_tag else 'N/A'
        
        # Extract the title
        title = listing.find('p', class_='truncate gr-p-title')
        title = title.text.strip() if title else 'N/A'

        # Extract details like location and time
        details = listing.find('p', class_='truncate gr-p-date-loc')
        details_text = details.text.strip() if details else 'N/A'

        # Split details to separate location and other info
        location = details_text.split('•')[-1].strip() if '•' in details_text else 'N/A'

        # Extract the image URL if available
        image_tag = listing.find('img', class_='lazy height_media ing_post_listing img_product bg_reload_img')
        image_url = image_tag.get('src', 'N/A') if image_tag else 'N/A'

        # Append extracted data to the list
        data.append([car_id, title, 'Car', 'Lexus', 'N/A', 'N/A', 'N/A', 'N/A', 'N/A', 'N/A', 'N/A', 'N/A', location, 'N/A', image_url])

    except Exception as e:
        print(f"Error processing a listing: {e}")

# Check if data is populated
if not data:
    print("No data extracted. Check your selectors or the website structure.")

# Define the CSV file name
csv_file = 'car_listings_updated.csv'

# Write the data to a CSV file
with open(csv_file, mode='w', newline='', encoding='utf-8') as file:
    writer = csv.writer(file)
    # Write the header
    writer.writerow([
        'Car_id', 'Title', 'Category', 'Brand', 'Model', 'Year', 'Tax_type', 
        'Condition', 'Body_type', 'Fuel', 'Transmission', 'Color', 'Location', 'Price', 'Image_URL'
    ])
    # Write the data
    writer.writerows(data)

print(f"Data has been saved to {csv_file}")
```
